## Paginated API

### What is paginated API call?
A paginated API splits large datasets into multiple pages and provides a way to fetch each page sequentially.

### We can extract data from Paginated APIS in four ways
1. Knowing how many pages/requests you will need
2. Doing what the API tells you
3. Going until there are no more results
4. Going until there is an error 

```python
## Pokemon API Example
## https://pokeapi.co/api/v2/pokemon?limit=100
```

```python
## import necessary libraries
import requests
import pandas as pd
import time
```

```python

## Set the initial URL for the API request
url = "https://pokeapi.co/api/v2/pokemon?limit=100"
## Make the API request and get the data from the first page 
response = requests.get(url)
## Parse the JSON response
data = response.json()
```

```python
limit = 100
offset = 0
total_pokemons = data['count']
all_pokemons = []

try:

    for i in range(offset, total_pokemons, limit):
        paginated_url = f"https://pokeapi.co/api/v2/pokemon?limit={limit}&offset={i}"
        print(f"Fetching data from: {paginated_url}")
        response = requests.get(paginated_url)
        data = response.json()
        all_pokemons.extend(data['results'])
        time.sleep(1)  # To avoid hitting the API rate limit
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
## Convert the list of all pokemons to a DataFrame
pokemons_df = pd.DataFrame(all_pokemons)
## Display the first 2 rows of the DataFrame
pokemons_df.head(2)
```

```python
pokemons_df.shape
```

```python
### Method: We know how many pages we have method
### Pros: Maybe easy to think about
### Cons: You need to know how many pages it is, not very
```

## Using to the API

```python
## URL for pagination approach
url = "https://pokeapi.co/api/v2/pokemon?limit=100"

## Initialize an empty list to store all pokemons
all_pokemons = []

##  Using to the API
while url:
    print(f"Fetching data from: {url}")
    response = requests.get(url)
    data = response.json()
    all_pokemons.extend(data['results'])
    url = data.get('next')
    time.sleep(1)  # To avoid hitting the API rate limit

## Convert the list of all pokemons to a DataFrame
pokemons_df = pd.DataFrame(all_pokemons)
```

```python
pokemons_df.head(1)
```

```python
### Method: Listening to the API
### Pros: The API knows going on inside of it, good if you don't know how many pages / results there are
### Cons: The API has to actually tell you the next page
```

## Going until there are no more results

```python
## 
pokemon = []
## 
offset = 0
limit = 100

## 
while True:
    print("-----------")
    url = f"https://pokeapi.co/api/v2/pokemon?offset={offset}&limit={limit}"
    print("Requesting", url)
    response = requests.get(url)
    data = response.json()

    if len(data['results']) == 0:
        print("There is no pages further")
        break

    pokemon.extend(data['results'])

    offset += 100
```

```python
pokemon_df = pd.DataFrame(pokemon)
```

```python
pokemon_df.head(2)
```

```python
### Method: Going until no more results
### Pros: Very flexible
### Cons: Extra requests, doesn't really listen to the API, more line of code
```

## Waiting until things break

```python
pokemon = []
offset = 0
limit = 100

while True:
    print("----------------")
    url = f"https://pokeapi.co/api/v2/pokemon?offset={offset}&limit={limit}"
    print('Requesting:', url)
    response = requests.get(url)
    data = response.json()

    try:
        print(data['results'][0])
    except Exception as e:
        print(e)
        print("There was an error! Exiting the loop")
        break

    pokemon.extend(data['results'])

    offset += 100
```

```python
pokemon_df = pd.DataFrame(pokemon)

print(pokemon_df.shape)

pokemon_df.tail(5)
```

```python
### Method: Wait intil you hit an error with your processing
### Pros: Super easy and flexible, good technique for scraping multiple pages
### Cons: What if there's some OTHER error that isn't a you've run out of data
```
