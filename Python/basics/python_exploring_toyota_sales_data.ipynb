```python
# Find a trnsaction value more than 30000 and car model sould be 'RAV4'
sales_data.loc[
                (sales_data['sale_amount'] > 30000) & (sales_data['car_model'] == 'RAV4'), 
               ['sale_rep_id', 'car_model', 'sale_amount', 'commission_value', 'sale_date_std']
               ]
```

```python
# Update commission_pct to 0.05 where carmodel is 'RAV4' and commission_pct is 0
sales_data.loc[
        (sales_data['car_model'] == 'RAV4') & 
        (sales_data['commission_pct'] == 0.0) & 
        (sales_data['sale_amount'] > 30000) &
        (sales_data['sale_status'] == 'Completed'),
        'commission_pct'
        ] = 0.05
```

## <strong>Accessing Data with iloc - Position-Based Indexing</strong>

. iloc is used to access the data frame by position instead of label
. Along with index value, we can also pass column positions

```python
# access the data frame using iloc 
# column and row by index position
sales_data.iloc[0:5, 0:5]
```

```python
# display the fifth row values using iloc
sales_data.iloc[4, 0:5]
```

## <strong>Comparing loc and iloc methods</strong>

```python
# loc: label based indexing
sales_data.loc[2:10, ['sale_rep_id', 'car_model', 'sale_amount', 'commission_value', 'sale_date_std']]
```

```python
# iloc: position based indexing
sales_data.iloc[2:10, [1, 3, 4, 8, 7]]
```

```python
# loc: label based indexing
# Select sales data, where sale amount is greater than 30000
sales_data.loc[sales_data[
    'sale_amount'] > 30000,
    ['sale_rep_id', 'car_model', 'sale_amount', 'commission_value', 'sale_date_std']
               ]
# iloc: position based indexing
```
