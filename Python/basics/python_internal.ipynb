#### Copy in the lists

1. Shallow copy
2. Deep copy

```python
lst1 = [[1, 2], [3, 4]]

print(lst1[0])
print()

print(lst1[0][1])
print()

print(lst1[1][1])
```

```python
lst2 = lst1

print(id(lst1))
print(id(lst2))
```

```python
# print id's of lst1 and lst2
print('object id of lst1 ',id(lst1))
print()
print('object id of lst2 ',id(lst2))
```

## <span style='color:grey'><strong>Shallow Copy</strong></span>

```python
# copy data lst1 data into lst3 using copy method
lst3 = lst1.copy()

print('Data from list objects:')
print('list 1 data :', lst1)
print('list 2 data :', lst2)
print('list 3 data :', lst3)
print()

print('object id of list 1 ', id(lst1))
print('object id of list 2 ', id(lst2))
print('object id of list 3 ', id(lst3))
```

```python
# append [7, 8] to lst1
lst1.append([7, 8])

print('Data from list objects:')
print('list 1 data :', lst1)
print('list 2 data :', lst2)
print('list 3 data :', lst3)
print()

print('object id of list 1 ', id(lst1))
print('object id of list 2 ', id(lst2))
print('object id of list 3 ', id(lst3))
```

```python
# append new list of values to lst3
# append [9, 10] to lst3
lst3.append([9, 10])

print('Data from list objects:')
print('list 1 data :', lst1)
print('list 2 data :', lst2)
print('list 3 data :', lst3)
print()

print('object id of list 1 ', id(lst1))
print('object id of list 2 ', id(lst2))
print('object id of list 3 ', id(lst3))
```

```python
# update one of the element value in lst1
lst1[2][0] = 300

print('Data from list objects:')
print('list 1 data :', lst1)
print('list 2 data :', lst2)
print('list 3 data :', lst3)
```

```python
# delete [300, 6] nested list from lst object
lst3.remove([300, 6])

print('Data from list objects:')
print('list 1 data :', lst1)
print('list 2 data :', lst2)
print('list 3 data :', lst3)
```

## <span style='color:grey'><strong>Deep Copy</strong></span>

```python
import copy
```

```python
lst3 = [[1, 2], [3, 4], [5, 6]]
lst4 = copy.deepcopy(lst3)

# Print data from lst3 and lst4
print('lst 3 data :', lst3)
print('lst 4 data :', lst4)
print()

# Print object ids of lst3 and lst4
print('object id of lst 3 :', id(lst3))
print('object id of lst 4 :', id(lst4))
```

```python
lst4[2][0] = 300

# Print data from lst3 and lst4
print('lst 3 data :', lst3)
print('lst 4 data :', lst4)
```
