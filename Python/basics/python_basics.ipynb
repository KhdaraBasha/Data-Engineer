#### SLICING
Slicing is used to access individual elements or a range of elements in a list

Python supports "slicing" indexable sequences. The syntax for slicing lists is:

        . list_object[start:end:step] 
        or
        . list_object[start:end]

start and end are indices (start inclusive, end inclusive). All slicing values are optional.

```python
l = list(range(10))
print(l)
```

```python
l[5:10]
```

```python
l[5:10:2]
```

```python
lst = list(range(10)) # Create a list with elements start from 0 and with total 10 elements
print(lst)
print()

print('elements from index between 4 to 7 :', lst[4:7])
print()

print('alternate elements, starting at index 0:', lst[0::2]) # Print alternate elements from the index 0 to till final index.
print()

print('every third element, starting at index 1:', lst[1::3]) # Print every third element from the index 1 to till final index
```

```python
# Simple program to find th even numbers in a list

# Initialized the list
lst = list(range(10))

for n in lst:                   # select one element in list_1
    if n % 2 == 0:              # check if it is even, only then goes to next stepelse performs above step
        print(f'{n}', end= ',')    # Prints n if even
```

#### Comprehensions

        . Python provides syntactic sugar to write small loops to generate lists/sets/tuples/dicts in one line
        . These are called comprehensions, and can greatly increase development speed and readability

Syntax:

        sequence = [expressions(element) for element in iterable if condition]

the brackets used for creating the comprehension define what type of object is created

Use [] for lists, () for generatorrs, {} for sets and dicts

#### List Comprehensions

```python
names = ['ravi', 'pooja', 'vijay', 'kiran']
hello = [ 'hello ' + name for name in names ]

print(hello)
```

```python
numbers = [55, 32, 87, 99, 10, 54, 32]

even_numbers = [ n for n in numbers if n % 2 == 0]
print(even_numbers)
print()

odd_numbers = [ (n, n * n) for n in numbers if n % 2 != 0]
print(odd_numbers)
```

#### EXCEPTION HANDLING

##### try and except
The basic terminology and syntax used to handle errors in python is the try and except statement.

Syntax:

        try:
            run this code
        except:
            execute this code when there is an exception
        else:
            no exception? run this code.
        finally:
            always run this code

```python
try:
    print(abcd)
except Exception as e:
    print(f'error {e}')
```

```python
try:
    x  = 0 / 5
except Exception as e:
    print('Execption Block')
    print(f'Error {e}')
    print('Executed when exception occurs')
    print()
else:
    print('Else Block')
    print('executed only when exception does not occur')
    print()
finally:
    print('Finally Block')
    print('finall block, always executed!.')
```

#### File I/O: Helps you read your files
        . Python provides a file object to read text/binary files
        . This is similar to to the FileStream object in other languages
        . Since file is a resource, it must be closed after use. This can be done manually, or using a context manager (with statement)

```python
# Create a file in the current directory
with open('data/myfile.txt', 'w') as f:
    f.write('This is my first file!\n')
    f.write('Second line!\n')
    f.write('Last line!\n')
    f.close
```

```python
with open('data/myfile.txt', 'r') as f:
    for line in f:
        print(line)

f.close()
```
