```python
## Importing Libraries
import pandas as pd
import numpy as np
import os
from datetime import datetime as dt
from pathlib import Path
import re
```

```python
## Load apple products data file into a DataFrame
df = pd.read_csv('data/apple_products.csv')
```

```python
## Display the first 5 rows of the DataFrame
df.head(5)
```

```python
## Display DataFrame information about columns, non-null counts, and data types
df.info()
```

```python
## Get the maximum value in the 'Mrp' column
df['Mrp'].max()
```

```python
## Get the minimum value in the 'Mrp' column
df['Mrp'].min()
```

```python
## Calculate the average price of iPhone products
df['Mrp'].mean()
```

```python
## Find the product details with the highest price
df[['Product Name', 'Mrp']][df['Mrp'] == df['Mrp'].max()]
```

```python
## Find the product details with the highest price
df.loc[df['Mrp'].eq(df['Mrp'].max()), ['Product Name', 'Mrp']]
```

```python
## Find the product details with the lowest price
df.loc[df['Mrp'].eq(df['Mrp'].min()),['Product Name', 'Product URL', 'Mrp', 'Sale Price']]
```

```python
## Find the product details with MRP greater than average price of all products
df.loc[df['Mrp'] >= np.round(df['Mrp'].mean()),['Product Name', 'Mrp', 'Sale Price']]
```

```python
## Extract Model Name from Product Name and create a new column 'Model Name'
df['Model Name'] = df['Product Name'].str[6:15].str.upper()
```

```python
df.head()
```

```python
## Extract Model Color from Product Name and create a new column 'Model Color'
df['Model Color'] = df['Product Name'].str.split('(').str[1].str.split(',').str[0]
```

```python
## Extract Model Storage from Product Name and create a new column 'Model Storage' (pattern-based, normalized)
df['Model Storage'] = df['Product Name'].astype(str).str.extract(r'(\d+\s?(?:GB|TB))', flags=re.IGNORECASE)[0].str.upper().str.replace(r'\s+','', regex=True)
```

```python
## Extrcat Earphodes Inclusion from Product Name and create a new column 'Earphones Included'
df['Earphones Included'] = np.where(df['Product Name'].str.contains('EarPods', case=False, na=False), 'Yes', 'No')
```

```python
## Extract Power Adapter Inclusion from Product Name and create a new column 'Power Adapter Included'
df['Power Adapter Included'] = np.where(df['Product Name'].str.contains('Power Adapter', case=False, na=False), 'Yes', 'No')
```
