## <span style='color:grey'><strong>Handling large CSV files in Chunks</strong></span>

```python
# Declaring chunk size to handle large datasets
chunk_size  = 1000

toyota_data_chunk_reader = pd.read_csv('data/toyota_sales_data.csv',
                                       chunksize= chunk_size)

type(toyota_data_chunk_reader)
```

```python
for toyota_chunk in toyota_data_chunk_reader:
    print(toyota_chunk.head(1))
```

## <span style='color:grey'><strong>Apply Transformations on DataFrame</strong></span>

```python
# Caluclate the commision amount. If value is null then fill with zero value and round it to two decimals
toyota_data['commission_amount'] = toyota_data['sale_amount'] * toyota_data['commission_pct'].fillna(0).round(2)
```
