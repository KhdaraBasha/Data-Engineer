#### Function Arguments
        . Parameters may or may not have default values (in the example below, 'argm' has default value 3)
        . A function parameter can be passed either by position or by key/name
        . A function can accept a variable number of positional arguments (*args)
        . A function can accept a variable number of keyword argumnets (**kwargs)
        . Function parameters are passed by object referenece

#### Default Arguments / Keyword Arguments
        . Default argumnets are those arguments which can be changed but their value remains the same if not given. 

```python
def double_number(n = 1):
    return n * 2

print(double_number())

print(double_number(5))

print(double_number(999999999))
```

#### Positional Arguments
        . Positional arguments are those which are not followed by an equal (=) sign and default value. In this we need to 
        provide an argument while calling the function, else we get an error. 

```python
def double_number(num):
    try:
        return num * 2
    except Exception as e:
        print(e)

print(double_number(4))

try:
    print(double_number())
except Exception as e:
    print(e)
```
