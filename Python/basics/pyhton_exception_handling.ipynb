<span style="color:black"><strong>SyntaxError</strong></span>:
Occurs when the Python interpreter encounters an error in the syntax of the code, preventing it from executing. This can include issues like incorrect indentation, missing parentheses, or invalid keywords.

<span style="color:black"><strong>IndentationError</strong></span>:
A specific type of SyntaxError that occurs when the indentation of code blocks is not consistent. Python uses indentation to define blocks of code, so incorrect or inconsistent indentation can lead to errors.

<span style="color:black"><strong>TypeError</strong></span>:
Occurs when an operation or function is applied to an object of inappropriate type. For example, performing arithmetic on non-numeric data types or passing arguments of the wrong type to a function.

<span style="color:black"><strong>NameError</strong></span>:
Occurs when a variable or function name used in the code is not defined or cannot be found in the current scope.

<span style="color:black"><strong>ValueError</strong></span>:
Occurs when a function receives an argument of the right type but inappropriate value. For example, passing an invalid argument to a function that expects a certain range of values.

<span style="color:black"><strong>IndexError</strong></span>:
Occurs when trying to access an element from a list, tuple, or string using an index that is out of its range. It indicates that the index provided does not exist in the sequence.

<span style="color:black"><strong>KeyError</strong></span>:
Occurs when trying to access a dictionary with a key that does not exist in the dictionary. It indicates that the specified key is not found in the dictionary.

<span style="color:black"><strong>AttributeError</strong></span>:
Occurs when trying to access an attribute or method that does not exist for a given object. It indicates that the object does not have the specified attribute or method.

<span style="color:black"><strong>ZeroDivisionError</strong></span>:
Occurs when attempting to divide a number by zero, which is mathematically undefined. It indicates that the denominator in a division operation is zero.

#### <span style="color:green"><strong> Exception Handling </strong></span>

We will exception handling method run the python code without stopping.

    try:
        code that might raise an exception
    except:
        code to handle the exception
    else:
        code to execute if no exception were raised
    finally:
        code to execute regardless of the exception


```python
try:
    num =int(input("Enter a number : "))
    result = 10 / num
    print(result)
except Exception  as e:
    print('Error generated')
    print(f'Error code is {e}')
```
