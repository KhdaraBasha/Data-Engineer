### <span style="color:grey"><strong>Lambda Function</strong></span>
Lambda Functions -Anonymous functions that can be used to define a single line of code

```python
## Create basic function 
def cube(x):
    return x * x * x

print(cube(3))
```

```python
## lambda method
lambda_fun = lambda x: x * x * x

lambda_fun(3)
```

#### Implement a code to add two numbers

```python
lambda_sum = lambda x,y: x + y

lambda_sum(3,5)
```

#### Sort a list of dictionaries on the basics of specific key

```python
emp_rec = [
    {'name' : 'John', 'age' : 29, 'occupation' : 'engineer'},
    {'name' : 'Vikash', 'age' : 25, 'occupation' : 'doctor'},
    {'name' : 'Nikhil', 'age' : 25, 'occupation' : 'engineer'},
    {'name' : 'Teja', 'age' : 23, 'occupation' : 'data scientist'},
    ]

emp_rec
```

```python
sorted(emp_rec, key = lambda x: x['age'] )
```

#### Sort the data on the basis of key value as age, name

```python
sorted(emp_rec, key = lambda x: (x['age'], x['name']))
```

### <span style='color:grey'><strong> Map Function </strong></span>

#### Implement a function which results the square of every number in the list

```python
# without lambda method
lst = [2, 3, 4, 5, 6, 7, 8, 9]

new_lst = []

for l in lst:
    new_lst.append(l*l)
print(new_lst)
```

```python
# with lambda methos
squ_lst = list(map(lambda lst : lst ** 2, lst))

squ_lst
```

```python
## Common elements between the lsts
l1 = [2, 5, 7, 9]
l2 = [2, 7]

# Expected output is [2, 7]

# Without functional tools
output1 = []
for i in l2:
    if i in l1:
        output1.append(i)
print(output1)
print()

# With functional tools

# List comprahension
output2 = [i for i in l2 if i in l1]
print(output2)
print()

# Lambda method
print(list(map(lambda l: l if l in l1 else None, l2)))
```

### <span style='color:grey'><strong>Filter</strong></span> 

Subset of the given list by having some sort of constraints

```python
# Return the even elements from the list
lst = [2, 3, 4, 5, 6, 7, 8, 9]
print('input list')
print(lst)
print()

# Even elements in the list
even_lst = list(filter(lambda x: x % 2 == 0, lst))
print('even elements')
print(even_lst)
print()

# Odd elements in the list
odd_lst = list(filter(lambda x: x % 2 != 0, lst))
print('odd elements')
print(odd_lst)
print()
```

### <span style='color:grey'><strong>Reduce</strong></span>
Given a list, return the sum of the entire list

Single output - reduce function

```python
# Return the sum of the entire elements from the list
lst  = [2, 3, 4, 5, 6, 7, 8]

# Without functional tools
tot_val  = 0

for i in lst:
    tot_val += i 
print('for loop method')
print(tot_val)

# With functional tools
from functools import reduce
print('reduce function')
tot_val1 = reduce(lambda x,y: x+y, lst)
print(tot_val1)
```
