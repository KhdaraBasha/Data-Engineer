## <span style='color:grey'><strong>Difference between the shallow copy and deep copy in numpy</strong></span>

```python
arr2
```

```python
shallowarr = arr2

deeparr = np.copy(arr2)
```

```python
# Print the data from arrays
print('array 2 :',arr2)
print()
print('Shallow array :', shallowarr)
print()
print('Deep array :', deeparr)
print()

# Print Object id of arrays
print('array 2 object id :',id(arr2))
print()
print('Shallow array object id:', id(shallowarr))
print()
print('Deep array object id:', id(deeparr))
```
